```python
# S-Situation: Setting up the environment and loading the dataset.
# T-Task: Import pandas and load the CSV file into a DataFrame.

import pandas as pd

try:
    df = pd.read_csv('/workspaces/machine_learning/car_fuel_efficiency.csv')
    print("Dataset loaded successfully.")
except FileNotFoundError:
    print("Error: The specified file 'car_fuel_efficiency.csv' was not found at the path.")
    df = pd.DataFrame() # Create an empty DataFrame to prevent errors.
```

```python
# A-Action: Check the number of records and unique fuel types.
# R-Result: Print the counts to verify the dataset's size and variety.

if not df.empty:
    records_count = len(df)
    fuel_types_count = df['fuel_type'].nunique()

    print(f"Total records in the dataset: {records_count}")
    print(f"Number of unique fuel types: {fuel_types_count}")
```

```python
# A-Action: Identify columns with any missing values.
# R-Result: Print the count of columns containing NaN values.

if not df.empty:
    cols_with_missing_values = df.columns[df.isnull().any()]
    num_cols_with_missing = len(cols_with_missing_values)

    print(f"Columns with missing values: {list(cols_with_missing_values)}")
    print(f"Number of columns with missing values: {num_cols_with_missing}")
```

```python
# A-Action: Filter for cars from 'Asia' and find the maximum 'mpg' value.
# R-Result: Display the highest fuel efficiency for the specified region.

if not df.empty:
    df_asia = df[df['origin'] == 'Asia'].copy() # Use .copy() to prevent SettingWithCopyWarning.
    max_fuel_efficiency_asia = df_asia['mpg'].max()

    print(f"Maximum fuel efficiency of cars from Asia: {max_fuel_efficiency_asia}")
```

```python
# A-Action: Calculate the median horsepower before and after filling NaN values.
# T-Task: Use the mode to impute missing 'horsepower' values and then re-calculate the median.

if not df.empty:
    # Calculate median before filling
    median_before = df['horsepower'].median()

    # Find the most frequent value (mode) and fill NaNs
    mode_horsepower = df['horsepower'].mode()[0]
    df_q6 = df.copy() # Work on a copy to not modify the original df
    df_q6['horsepower'].fillna(mode_horsepower, inplace=True)

    # Recalculate the median after filling
    median_after = df_q6['horsepower'].median()

    print(f"Median of horsepower (before filling NaN): {median_before}")
    print(f"Median of horsepower (after filling NaN with mode): {median_after}")

    # R-Result: Check if the median has changed.
    change = "changed" if median_after != median_before else "not changed"
    print(f"Result: The median has {change}.")
```

```python
# S-Situation: Perform a linear regression calculation on a subset of the data.
# A-Action: Select data, perform matrix multiplications, and calculate the sum of regression weights (w).
# R-Result: Print the final sum of the calculated weights.

import numpy as np

if not df.empty:
    df_asia_q7 = df[df['origin'] == 'Asia']

    # Select the first 7 rows of specific columns
    X_df = df_asia_q7[['vehicle_weight', 'model_year']].head(7)
    X = X_df.to_numpy()

    # Predefined y array for the calculation
    y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])

    # Calculate w using the formula: w = (X^T * X)^-1 * X^T * y
    XTX = X.T @ X
    XTX_inv = np.linalg.inv(XTX)
    w = XTX_inv @ X.T @ y
    w_sum = np.sum(w)

    print(f"Selected data (X) for the calculation:\n{X_df}")
    print(f"\nThe sum of all elements of w is: {w_sum:.3f}")
```
